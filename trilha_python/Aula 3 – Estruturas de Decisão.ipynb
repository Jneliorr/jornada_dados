### Exercício 2 – Verificando os tipos ###


Considere as variáveis:

numero_camisa = 10

jogador = "Raphinha"

Utilize isinstance() para verificar:

se numero_camisa é um int;
se numero_camisa é uma str;
se jogador é uma str.

Mostre o resultado de cada verificação na tela.


```python
numero_camisa = 10
jogador = "Raphina"

if isinstance( numero_camisa, int):
    print(f'A variavel numero camisa é um tipo inteiro coma informação: {numero_camisa}')
else:
    print(f'A variavel numero_camisa não é do tipo inteiro ela é: {type(numero_camisa)}, com a informação de: {numero_camisa}')
if isinstance( jogador, str):
    print(f'A variave jogador é um tipo texto e é: {jogador}')
else:
    print(f'A variavel jogador não é do tipo int ela é: {type(jogador)}, com a informação de: {jogador}')
```
