### TUPLA

```python
partida = (
    "Brasil",
    "México",
    "Estádio Central",
    "20:00"
)
```

```python
print(partida[0])  # Brasil
print(partida[-1])  # 20:00
```

```python
# Desempacotamento
selecao = partida[0]
adversario = partida[1]
estadio = partida[2]
horario = partida[3]

print(selecao, adversario, estadio, horario)
```

```python
print (f'{selecao} x {adversario} | {estadio} | {horario}')
```

### SET

```python
jogadores_que_fizeram_gol = set(autores_gols)

print(len(jogadores_que_fizeram_gol), jogadores_que_fizeram_gol)
```

### DICIONARIOS

```python
gols_jogador = {

}



for jogador in autores_gols:
    if jogador in gols_jogador:
        gols_jogador[jogador] += 1
    else:
        gols_jogador[jogador] = 1

print(gols_jogador)
```

```python
for jogador, gols in gols_jogador.items():
    print(f'jogador: {jogador} | gols: {gols}')
```
