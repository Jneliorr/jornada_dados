```python
# Exercício 1 – Cadastro de Figurinhas

# Você foi contratado para desenvolver um sistema simples para controlar um álbum de figurinhas da Copa do Mundo.

# O programa deve permitir que o usuário cadastre figurinhas até que ele digite a palavra "fim".

# Ao final, exiba:

# Todas as figurinhas cadastradas.
# A quantidade total de figurinhas.
# A primeira figurinha cadastrada.
# A última figurinha cadastrada.

# Desafio: não permita que o usuário cadastre uma figurinha vazia.

```

```python
figurinhas = []
while True:
    figurinha = input("Digite o nome da figurinha (ou 'fim' para encerrar): ")
    if figurinha.lower() == "fim":
        break
    elif figurinha.strip() == "":
        print("Figurinha vazia não é permitida. Tente novamente.")
        continue
    figurinhas.append(figurinha)
print("Cadastro de figurinhas encerrado.")
print(f"Total de figurinhas cadastradas: {len(figurinhas)}")
print(f"Primeira figurinha cadastrada: {figurinhas[0]}")
print(f"Última figurinha cadastrada: {figurinhas[-1]}")
print(f"Figurinhas cadastradas: {figurinhas}")
```

```python
# Peça ao usuário para informar o nome de 8 seleções.

# Armazene essas seleções em um set.

# Ao final:

# Exiba todas as seleções classificadas.
# Informe quantas seleções diferentes foram cadastradas.

# Depois pergunte ao usuário o nome de uma seleção e informe se ela está classificada utilizando o operador in.

# Desafio: explique por que, mesmo digitando uma seleção repetida, ela aparece apenas uma vez no conjunto.
```

```python
while True:
    selecoes = set()
    for i in range(8):
        selecao = input(f"Digite o nome da seleção {i + 1}: ")
        selecoes.add(selecao)

    print("\nSeleções classificadas:")
    for selecao in sorted(selecoes):
        print(selecao)

    print(f"\nTotal de seleções diferentes cadastradas: {len(selecoes)}")

    selecao_verificar = input("\nDigite o nome de uma seleção para verificar se está classificada: ")
    if selecao_verificar in selecoes:
        print(f"A seleção {selecao_verificar} está classificada.")
    else:
        print(f"A seleção {selecao_verificar} não está classificada.")

    break
```

```python
# Exercício 4 – Menu do Álbum

# Desenvolva um programa que simule um álbum de figurinhas.

# Utilize uma lista para armazenar as figurinhas e exiba o seguinte menu:

# 1 - Adicionar figurinha
# 2 - Remover figurinha
# 3 - Buscar figurinha
# 4 - Mostrar álbum
# 5 - Encerrar

# O menu deve permanecer sendo exibido até que o usuário escolha a opção 5.

# Regras:

# Ao adicionar, a figurinha deve ser inserida no final da lista.
# Ao remover, informe caso a figurinha não exista.
# Na busca, informe se a figurinha está ou não no álbum.
# Ao mostrar o álbum, exiba todas as figurinhas utilizando um for.

# Desafio: utilize if, elif, else e while.

```

```python
while True:
    print("\nMenu do Álbum de Figurinhas:")
    print("1 - Adicionar figurinha")
    print("2 - Remover figurinha")
    print("3 - Buscar figurinha")
    print("4 - Mostrar álbum")
    print("5 - Encerrar")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        nova_figurinha = input("Digite o nome da figurinha para adicionar: ")
        if nova_figurinha.strip() == "":
            print("Figurinha vazia não é permitida. Tente novamente.")
        else:
            figurinhas.append(nova_figurinha)
            print(f"Figurinha '{nova_figurinha}' adicionada ao álbum.")

    elif opcao == "2":
        figurinha_remover = input("Digite o nome da figurinha para remover: ")
        if figurinha_remover in figurinhas:
            figurinhas.remove(figurinha_remover)
            print(f"Figurinha '{figurinha_remover}' removida do álbum.")
        else:
            print(f"Figurinha '{figurinha_remover}' não encontrada no álbum.")

    elif opcao == "3":
        figurinha_buscar = input("Digite o nome da figurinha para buscar: ")
        if figurinha_buscar in figurinhas:
            print(f"Figurinha '{figurinha_buscar}' está no álbum.")
        else:
            print(f"Figurinha '{figurinha_buscar}' não está no álbum.")

    elif opcao == "4":
        if figurinhas:
            print("\nÁlbum de Figurinhas:")
            for idx, figurinha in enumerate(figurinhas, start=1):
                print(f"{idx}. {figurinha}")
        else:
            print("O álbum está vazio.")

    elif opcao == "5":
        print("Encerrando o programa. Até logo!")
        break

    else:
        print("Opção inválida. Tente novamente.")
```

```python
# Exercício 6 – Desafio Final: Sistema da Copa do Mundo

# Desenvolva um programa para cadastrar partidas da Copa do Mundo.

# O programa deverá permanecer em execução até que o usuário decida encerrá-lo.

# Para cada partida, solicite:

# Seleção mandante
# Seleção visitante
# Gols da seleção mandante
# Gols da seleção visitante

# Cada partida deverá ser armazenada em um dicionário, e todos os dicionários deverão ser armazenados em uma lista.

# Ao finalizar o cadastro, exiba:

# A quantidade de partidas cadastradas.
# Todas as partidas registradas.
# Quantas partidas terminaram empatadas.
# A partida com o maior número total de gols.
# A média de gols por partida.

# Requisitos:

# Utilize listas e dicionários.
# Utilize while para controlar o cadastro.
# Utilize for para percorrer as partidas.
# Utilize if, elif e else para identificar o resultado de cada jogo.
# Utilize try/except para validar os gols informados.
# Utilize len() para calcular a quantidade de partidas cadastradas.



    
```

```python
partidas = []

while True:
    resposta = input(
        "\nDeseja cadastrar uma partida? "
        "(s para sim, n para finalizar): "
    ).strip().lower()

    if resposta == "n":
        break

    elif resposta != "s":
        print("Opção inválida. Digite 's' ou 'n'.")
        continue

    selecao_mandante = input("Seleção mandante: ").strip()
    selecao_visitante = input("Seleção visitante: ").strip()

    while True:
        try:
            gols_mandante = int(
                input(f"Gols da seleção {selecao_mandante}: ")
            )

            if gols_mandante < 0:
                print("A quantidade de gols não pode ser negativa.")
                continue

            break

        except ValueError:
            print("Digite uma quantidade inteira válida.")

    while True:
        try:
            gols_visitante = int(
                input(f"Gols da seleção {selecao_visitante}: ")
            )

            if gols_visitante < 0:
                print("A quantidade de gols não pode ser negativa.")
                continue

            break

        except ValueError:
            print("Digite uma quantidade inteira válida.")

    partida = {
        "mandante": selecao_mandante,
        "visitante": selecao_visitante,
        "gols_mandante": gols_mandante,
        "gols_visitante": gols_visitante
    }

    partidas.append(partida)
    print("Partida cadastrada com sucesso!")


print("\n===== RESUMO DAS PARTIDAS =====")

quantidade_partidas = len(partidas)
print(f"Quantidade de partidas cadastradas: {quantidade_partidas}")

if quantidade_partidas == 0:
    print("Nenhuma partida foi cadastrada.")

else:
    empates = 0
    partida_maior_numero_gols = partidas[0]
    total_gols = 0

    print("\nPartidas registradas:")

    for partida in partidas:
        mandante = partida["mandante"]
        visitante = partida["visitante"]
        gols_mandante = partida["gols_mandante"]
        gols_visitante = partida["gols_visitante"]

        print(
            f"{mandante} {gols_mandante} x "
            f"{gols_visitante} {visitante}"
        )

        if gols_mandante == gols_visitante:
            empates += 1
            resultado = "Empate"

        elif gols_mandante > gols_visitante:
            resultado = f"Vitória do {mandante}"

        else:
            resultado = f"Vitória do {visitante}"

        print(f"Resultado: {resultado}")

        gols_da_partida = gols_mandante + gols_visitante
        total_gols += gols_da_partida

        maior_total = (
            partida_maior_numero_gols["gols_mandante"]
            + partida_maior_numero_gols["gols_visitante"]
        )

        if gols_da_partida > maior_total:
            partida_maior_numero_gols = partida

    maior_total = (
        partida_maior_numero_gols["gols_mandante"]
        + partida_maior_numero_gols["gols_visitante"]
    )

    media_gols = total_gols / len(partidas)

    print(f"\nQuantidade de partidas empatadas: {empates}")

    print(
        "Partida com maior número total de gols: "
        f"{partida_maior_numero_gols['mandante']} "
        f"{partida_maior_numero_gols['gols_mandante']} x "
        f"{partida_maior_numero_gols['gols_visitante']} "
        f"{partida_maior_numero_gols['visitante']} "
        f"({maior_total} gols)"
    )

    print(f"Média de gols por partida: {media_gols:.2f}")
```
